```python
import pandas as pd
import numpy as np
from statistics import mean, stdev
from sklearn.naive_bayes import GaussianNB
from sklearn import svm
from sklearn.metrics import f1_score, fbeta_score, roc_auc_score
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn import preprocessing
from sklearn.model_selection import StratifiedKFold
from sklearn import linear_model
#reda labels
label=np.genfromtxt('S1.3.labels.txt')
label=label[1:len(label)]-1
#for full data
#data=  pd.read_csv('S1.'+'3'+'.csv',index_col=0)
##for reduced data
data=pd.read_csv('reduced_data_S1.1.3.csv',index_col=0)

data=data.to_numpy()

x = data                        
 
# Input_ y_Target_Variable.
y = label                     
   
# Feature Scaling for input features.
scaler = preprocessing.MinMaxScaler()
x_scaled = scaler.fit_transform(x)

#lr = linear_model.LogisticRegression()
clf = RandomForestClassifier(max_depth=2, random_state=0) # GradientBoostingClassifier(n_estimators=100, learning_rate=1.0,max_depth=1, random_state=0) # GaussianNB()  
# Create StratifiedKFold object.
skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=1)
lst_accu_stratified = []
lst_f1_stratified = []


#clf_accu_stratified  
for train_index, test_index in skf.split(x, y):
    x_train_fold, x_test_fold = x_scaled[train_index], x_scaled[test_index]
    y_train_fold, y_test_fold = y[train_index], y[test_index]
    clf.fit(x_train_fold, y_train_fold)
    #lst_accu_stratified.append(clf.score(x_test_fold, y_test_fold))
    y_pred_fold = clf.predict(x_test_fold)
   # lst_f1_stratified.append(fbeta_score(y_test_fold, y_pred_fold,average='weighted', beta=2))
    lst_f1_stratified.append(roc_auc_score(y_test_fold, y_pred_fold))
# Print the output.
print('List of possible f1_score:', lst_f1_stratified)
print('\nMaximum f1_score That can be obtained from this model is:',
      max(lst_f1_stratified)*100, '%')
print('\nMinimum f1_score:',
      min(lst_f1_stratified)*100, '%')
print('\nOverall f1_score:',
      mean(lst_f1_stratified)*100, '%')
print('\nStandard Deviation is:', stdev(lst_f1_stratified))
```
